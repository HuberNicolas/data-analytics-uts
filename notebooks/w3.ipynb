```python
# Nominal, Ordinal, Interval, Ratio
# Variable classifications in the Iris dataset

# ID (Row): Identifier - Uniquely identifies each record, not used for analysis.

# The "Row" or "ID" attribute is typically used to uniquely identify each record in the dataset and does not contribute to the actual data analysis.
# It is a categorical variable, but more specifically, it's an identifier variable.
# Identifier variables are a subtype of categorical variables used exclusively for identification and not for analytical comparisons or measurements.
print("Row (ID): Identifier")

# Sepal Length: Ratio - Has a true zero and equal intervals, suitable for mathematical operations.
print("Sepal Length: Ratio")

# Sepal Width: Ratio - Has a true zero and equal intervals, allowing for ratio comparisons.
print("Sepal Width: Ratio")

# Petal Length: Ratio - Similar to Sepal Length and Width, allows for full quantitative analysis.
print("Petal Length: Ratio")

# Petal Width: Ratio - Allows for meaningful ratio comparisons and quantitative analysis.
print("Petal Width: Ratio")

# Species: Nominal - Categorizes data without implying any order or measurement scale.
print("Species: Nominal")
```

```python
def summary(df):
    print("INFO: ")
    print(df.info())
    
    print("DESCRIPTION: ")
    print(df.describe(include='all'))
    
    print("TYPES: ")
    print(df.dtypes)
```

```python
df_imports = pd.read_csv(filepath_or_buffer="../workshops/w3/imports-85-1.csv")
summary(df=df_imports)
```
